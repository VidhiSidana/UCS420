```python
import numpy as np
x = np.array([1, 2, 3, 4, 5, 1, 2, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(x, return_counts=True)
most_frequent_x = values[np.argmax(counts)]
indices_x = np.where(x == most_frequent_x)[0]

print(x)
print("Most frequent value in x:", most_frequent_x)
print("Indices:", indices_x)
values, counts = np.unique(y, return_counts=True)
most_frequent_y = values[np.argmax(counts)]
indices_y = np.where(y == most_frequent_y)[0]
print(y)
print("Most frequent value in y:", most_frequent_y)
print("Indices:", indices_y)
```

```python
import numpy as np
ucs420_vidhi = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_vidhi)
print("\nMean:", np.mean(ucs420_vidhi))
print("Median:", np.median(ucs420_vidhi))
print("Maximum:", np.max(ucs420_vidhi))
print("Minimum:", np.min(ucs420_vidhi))
resized_ucs420_vidhi = ucs420_vidhi.reshape(4, 3)
print(resized_ucs420_vidhi)
```
