```python
import random
from collections import Counter
```

```python
random.seed(roll_no)
random_numbers=[random.randint(100,900) for _ in range(900)]
print(random_numbers)
```

```python
odd_numbers=[x for x in random_numbers if x%2!=0]
print(odd_numbers)
```

```python
even_numbers=[x for x in random_numbers if x%2==0]
print(even_numbers)
```

```python
def is_prime(n):
  if n<2:
    return False
  for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
  return True
```

```python
prime_numbers=[x for x in random_numbers if is_prime(x)]
print(len(prime_numbers))
print(prime_numbers)
```

```python
frequency = Counter(random_numbers)

highest_frequency, frequency_count = frequency.most_common(1)[0]

print(highest_frequency)
print(frequency_count)
```

```python
digits=[int(d) for d in roll_no]
A={digit*7 for digit in digits}
B={digit*9 for digit in digits}
print(A)
print(B)
```

```python
union=A.union(B)
print(union)
```

```python
intersection=A.intersection(B)
print(intersection)
```

```python
A_minus_B=A.difference(B)
B_minus_A=B.difference(A)
print(A_minus_B)
print(B_minus_A)
```

```python
symmetric_difference=A.symmetric_difference(B)
print(symmetric_difference)
```

```python
print(A.issubset(B))
print(B.issuperset(A))
```
