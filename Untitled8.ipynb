```python
import pandas as pd

data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01',
                     '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

print(employees)
```

```python
employees.to_csv('employees.csv', index=False)
```

```python
df = pd.read_csv('employees.csv')

print(df)
```

```python
print("Shape of DataFrame:", df.shape)
```

```python
df.info()
```

```python
print(df.describe())
```

```python
print("First 5 rows:")
print(df.head(5))

print("\nLast 3 rows:")
print(df.tail(3))
```

```python
average_salary = df['Salary'].mean()

print("Average Salary:", average_salary)
```

```python
total_bonus = df['Bonus'].sum()

print("Total Bonus:", total_bonus)
```

```python
youngest_age = df['Age'].min()

print("Youngest Employee Age:", youngest_age)
```

```python
highest_rating = df['Rating'].max()

print("Highest Performance Rating:", highest_rating)
```

```python
df_sorted = df.sort_values(by='Salary', ascending=False)

print(df_sorted)
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

df['Performance'] = df['Rating'].apply(performance_category)

print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df.rename(columns={'Employee_ID': 'ID'}, inplace=True)

print(df)
```

```python
employees_experience = df[df['Years_of_Experience'] > 5]

print(employees_experience)
```

```python
it_employees = df[df['Department'] == 'IT']

print(it_employees)
```

```python
df['Tax'] = df['Salary'] * 0.10

print(df)
```

```python
df['Salary_After_Tax'] = df['Salary'] * 0.90
```

```python
df.to_csv('modified_employees.csv', index=False)

print("Modified DataFrame saved successfully!")
```
