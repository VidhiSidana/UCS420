```python
n = int(input("Enter a number: "))

sum = 0

for i in range(2, n + 1):
    prime = True

    for j in range(2, i):
        if i % j == 0:
            prime = False
            break

    if prime:
        sum = sum + i

print("Sum of prime numbers =", sum)
```

```python
def addodd(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total

n = int(input("Enter number: "))
print(addodd(n))
```

```python
def addprime(n):
    total = 0

    for i in range(2, n + 1):
        prime = True

        for j in range(2, i):
            if i % j == 0:
                prime = False
                break

        if prime:
            total += i

    return total

n = int(input("Enter number: "))
print(addprime(n))
```
